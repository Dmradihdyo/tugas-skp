```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Load the dataset
df = pd.read_csv('fake_news_dataset.csv')

# Check if data loaded correctly
print(df.head())

# Split the data into features and target
X = df['Headline']
y = df['Label']

# Convert text data to TF-IDF features
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(X)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.3, random_state=42)

# Initialize and train the Logistic Regression model
lr_model = LogisticRegression(max_iter=200)
lr_model.fit(X_train, y_train)

# Predict and evaluate the model
y_pred = lr_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'Logistic Regression Accuracy: {accuracy:.2f}')

# Test the model with new headlines
test_headlines = [
    'Breaking: New Technology Promises to Revolutionize Energy Sector',
    'Fake News: Politician Involved in Major Scandal'
]

# Convert new headlines to TF-IDF features
test_headlines_tfidf = vectorizer.transform(test_headlines)

# Predict labels for new headlines
predictions = lr_model.predict(test_headlines_tfidf)

# Display the predictions
for headline, prediction in zip(test_headlines, predictions):
    print(f'Headline: {headline}\nPrediction: {prediction}\n')
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import ipywidgets as widgets
from IPython.display import display

# Step 1: Create the Dataset
data = {
    'Headline': [
        'Breaking News: Local Mayor Found Guilty of Corruption',
        'Scientists Discover Cure for Common Cold',
        'Celebrity Donates Millions to Charity',
        'Aliens Invade Earth, World Leaders in Panic',
        'New Study Shows Chocolate Can Help You Lose Weight',
        'Man Claims to Have Time Traveled Back to the Future',
        'Government Announces Free Healthcare for All Citizens',
        'Fake News: Actor Involved in Major Scandal',
        'Experts Predict Economic Boom in Next Year',
        'Fake News: Giant Spider Found in Local Park'
    ],
    'Label': ['real', 'fake', 'real', 'fake', 'fake', 'fake', 'real', 'fake', 'real', 'fake']
}

# Create DataFrame
df = pd.DataFrame(data)

# Step 2: Prepare the Data and Train the Model
# Load the dataset (no need to save to CSV in this example)
# Split the data into features and target
X = df['Headline']
y = df['Label']

# Convert text data to TF-IDF features
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(X)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.3, random_state=42)

# Initialize and train the SVM model
svm_model = SVC()
svm_model.fit(X_train, y_train)

# Predict and evaluate the model
y_pred = svm_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'SVM Accuracy: {accuracy:.2f}')

# Step 3: Interactive Prediction Widget
text_input = widgets.Text(
    value='',
    placeholder='Headline',
    description='Message:',
    disabled=False
)
display(text_input)

button = widgets.Button(description="Predict")
output = widgets.Output()

display(button, output)

def on_button_clicked(b):
    with output:
        output.clear_output()  # Clear previous output
        message = text_input.value
        message_tfidf = vectorizer.transform([message])  # Transform the input message
        prediction = svm_model.predict(message_tfidf)[0]  # Predict the label
        print(f'Message: {message}\nPrediction: {prediction}\n')

button.on_click(on_button_clicked)
```
